```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/Users/premthatikonda/Downloads/FeatureEnggAndMLOPs/data/raw/Unit-02/daily_enrollments.csv')
df = df.sort_values("date").reset_index(drop=True)
print(df.shape)
df.head()
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df["date"], df["new_enrollments"], color='blue', label='Enrollments')
plt.xlabel('Date')
plt.ylabel('Enrollments')
plt.title('Daily Enrollments Over Time')
plt.legend()
plt.show()
```
